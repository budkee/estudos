# Estatística e Distribuição de Dados

## Conjunto de análise

```python
# Imports
import random
import statistics as st
import numpy as np
from scipy import stats
import pandas as pd
```

```python
# Criando o conjunto de dados pseudo-aleatórios
inicio_intervalo = 1
fim_intervalo = 100
tam_conjunto = 50
conjunto = list(map(lambda _: random.randint(inicio_intervalo, fim_intervalo), range(tam_conjunto)))
```

```python
# Conjunto desordenado
print(conjunto)
```

```python
# Conjunto ordenado
print(sorted(conjunto))
```

```python
# Sobre o conjunto segundo SciPy
stats.describe(conjunto)
```

## Medidas de Tendência Central ou de Posição

### Média

É o valor médio de um conjunto obtido a partir da soma dos valores dividido pelo tamanho do conjunto. 

```python
# Média com o módulo built-in do python
st.mean(conjunto)
```

```python
# Média com numPy
np.mean(conjunto)
```

### Mediana

É o valor que está no meio do conjunto, quando ordenado de forma crescente e o tamanho do conjunto é ímpar. Quando o tamanho do conjunto for par, deve-se calcular a média dos dois valores centrais e este será a mediana do conjunto. 

```python
# Mediana com statistics
st.median(conjunto)
```

```python
# Mediana com numPy
np.median(conjunto)
```

### Moda

É o valor que mais se repete no conjunto. Não há uma função que realize este cálculo no numPy.

```python
# Moda com statistics
st.mode(conjunto)
```

```python
# Moda com sciPy
stats.mode(conjunto, keepdims=False)
```

## Medidas Separatrizes

Assim como foi necessário ordenar de modo crescente o conjunto de dados para calcular a mediana, que por sua vez também é um tipo de medida separatriz, a coleta dos valores separatrizes também deve ter como pré-requisito a ordenação crescente dos dados.

### Quartis

Divide a distribuição em 4 partes iguais e coleta o valor representativo do respectivo quartil (1/4). Assim, num conjunto de dados serão coletados 3 valores que irão marcar a subdivisão dessas partes. 

### Decis

Divide a distribuição em 10 partes iguais e coleta o valor representativo do respectivo decil (1/10).

### Centis

Divide a distribuição em 100 partes iguais e coleta o valor representativo do respectivo centil (1/100).

## Diagrama de Dispersão

Um diagrama de dispersão é a representação gráfica (cartesiana) de uma análise envolvendo dados relativos a um contexto, onde se relacionam duas variáveis (uma independente, geralmente x, e outra dependente, geralmente y) e os dados de cada registro é representado por pontos. Ele é útil para visualizar de forma rápida a relação, ou a correlação, e frequência de resultados daquele determinado contexto, além de ser fácil de apontar os outliers e padrões.
