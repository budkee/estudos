## Fatura Inter 

```python
inter_01 = pd.read_csv('fatura-inter-2025-01.csv')
inter_01.info()
```

### Convertendo o tipo da coluna de valores

```python
inter_01['Valor']

# Remover o "R$" e o espaço
inter_01['Valor'] = inter_01['Valor'].str.replace(r'R\$\s*', '', regex=True)

# Remover os pontos usados como separadores de milhares
inter_01['Valor'] = inter_01['Valor'].str.replace(r'\.', '', regex=True)

# Substituir a vírgula por ponto e converter para float
inter_01['Valor'] = inter_01['Valor'].str.replace(',', '.').astype(float)
```

```python
categorias_inter_01 = inter_01.groupby('Categoria')['Valor'].sum()
categorias_inter_01
```
