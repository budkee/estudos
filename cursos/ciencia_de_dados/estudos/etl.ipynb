# Cheatsheet de Tratamento de Dados

### Limpa linha

```python
linha = '"Robert Downey, Jr.",3947.30 ,53,74.50 ,The Avengers,623.40'
print("Antes: \n", linha)
# Separa os valores ao encontrar uma vírgula e coloca em um vetor
linha = linha.split(',')
print("Problema: \n", linha)

# Concatenar os dois primeiros valores do vetor e ajustar a lista
linha[0] = linha[0] + linha[1]
linha[0] = linha[0].strip('"')
# print(linha)

# Deletar o elemento concatenado
del linha[1]
linha = ', '.join(linha)
print("Depois: \n", linha)
```

## Extração e Armazenamento de Dados

#### Extração 01

```python
# Abrir, ler e recolher os dados e fechar o arquivo
arquivo = open('../data/.csv/nomes_small.csv')
dados = arquivo.read()
arquivo.close()

# Visualizar os dados
for registro in dados.splitlines():
    print(*registro.split(','))
```

#### Extração 02
- Uso do `open` e `close` por *stream*

```python
# Abrir o arquivo
arquivo = open('../data/.csv/nomes_small.csv')

# Visualizar os dados
for registro in arquivo:
    print(*registro.split(','))

# Fechar o arquivo
arquivo.close()
```

#### Extração 03
- Uso do `open` e `close` por *stream*
- Neste caso, a diferença é a adição do `strip()` que remove espaços e outros delimitadores de linhas que a leitura de um *iterator* traz

```python
# Abrir o arquivo
arquivo = open('../data/.csv/nomes_small.csv')

# Visualizar os dados
for registro in arquivo:
    print(*registro.strip().split(','))

# Fechar o arquivo
arquivo.close()
```

#### Extração 04
- Uso do `try` e `finally`
- Uma opção mais robusta e segura durante a manipulação de dados

```python
# Abrir o arquivo
arquivo = open('../data/.csv/nomes_small.csv')

# Visualizar os dados
try:
    for registro in arquivo:
        print(*registro.strip().split(','))

# Fechar o arquivo
finally:
    arquivo.close()

if arquivo.closed:
    print('\nArquivo fechado com sucesso!')
```

#### Extração 05
- Uso do `with ... as`
- Uma opção ainda mais segura durante a manipulação de dados que, ao entrar no bloco, simula a execução do `try ... finally` dentro do contexto daquele bloco.

```python
# Abrir, ler e recolher os dados e fechar o arquivo
with open('../data/.csv/nomes_small.csv') as arquivo:
    for registro in arquivo:
        print(*registro.strip().split(','))
```

#### Extração com `csv`

```python
import csv

# Abrir, ler e recolher os dados e fechar o arquivo
with open('../data/.csv/nomes_small.csv') as arquivo:
    for registro in csv.reader(arquivo):
        print(*registro)
```

#### Gravação em arquivos externos

```python
# Abrir, ler e recolher os dados e fechar o arquivo
with open('../data/.csv/nomes_small.csv') as arquivo:
    with open('saidas/nomes_small.txt', 'w') as saida:
        for registro in arquivo:
            nome = registro.strip().split(',')
            print(*nome, file=saida)
```

##### Configuração e Conexão | PostgreSQL

```python
import os
DEBUG = True
DATABASE_URI = f"postgresql+psycopg2://{os.getenv('POSTGRES_USER')}:{os.getenv('POSTGRES_PASSWORD')}@{os.getenv('POSTGRES_HOST')}:{os.getenv('POSTGRES_PORT')}/nomes"

# ---------------- Objetos para comunicação com o PostgreSQL ----------------
engine = create_engine(DATABASE_URI)
Session = sessionmaker(autocommit=False, autoflush=False, bind=engine)
metadata = MetaData()
Base = declarative_base(metadata=metadata)
```

##### Configuração e Conexão | MySQL

```python
from mysql.connector import connect

conexao = connect(
    host=os.getenv('MYSQL_HOST'),
    port=os.getenv('MYSQL_PORT'),
    user='root',
    passwd=os.getenv('MYSQL_ROOT_PASSWORD')
)
# print(conexao)
cursor = conexao.cursor()
cursor.execute('CREATE DATABASE teste')
```

##### Listar BDs | MySQL

```python
cursor = conexao.cursor()
cursor.execute('SHOW DATABASES')
for i, database in enumerate(cursor, start=1):
    print(f'Banco de Dados {i}: {database[0]}')
```
